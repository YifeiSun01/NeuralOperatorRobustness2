```python
import torch, pickle, os

# 找到当前目录下的唯一 pt 和 pickle 文件
files = os.listdir(".")
pt_file   = [f for f in files if f.endswith(".pt")]
pkl_file  = [f for f in files if f.endswith((".pkl", ".pickle"))]

assert len(pt_file) == 1, f"需要唯一的 .pt 文件，找到: {pt_file}"
assert len(pkl_file) == 1, f"需要唯一的 .pkl/.pickle 文件，找到: {pkl_file}"

pt_file, pkl_file = pt_file[0], pkl_file[0]
print(pt_file, pkl_file)

# 加载
pt_obj  = torch.load(pt_file, map_location="cpu", weights_only=False)
with open(pkl_file, "rb") as f:
    pkl_obj = pickle.load(f)

def summarize(obj, name="", indent=0):
    pad = "  " * indent
    if isinstance(obj, dict):
        print(f"{pad}{name} dict with {len(obj)} keys: {list(obj.keys())[:10]}")
        for k,v in list(obj.items())[:5]:
            summarize(v, name=f"[{k}]", indent=indent+1)
    elif isinstance(obj, (list, tuple)):
        print(f"{pad}{name} {type(obj).__name__} len={len(obj)}")
        for i,v in enumerate(obj[:5]):
            summarize(v, name=f"[{i}]", indent=indent+1)
    elif isinstance(obj, torch.Tensor):
        print(f"{pad}{name} Tensor shape={tuple(obj.shape)} dtype={obj.dtype}")
    else:
        print(f"{pad}{name} {type(obj).__name__}")

print("=== .pt file structure ===")
summarize(pt_obj)

print("\n=== .pkl/.pickle file structure ===")
summarize(pkl_obj)
```

```python
import torch, pickle, os
import numpy as np

# 原始文件名
pt_file  = "dim1d_nx1024_N1500_solver=exponax_kernel=gaussian_correlation_length0.03_bcperiodic_nu0.0005_t1.0_seed45.pt"
pkl_file = "gaussiandiffdistr_cl0.03_dim=1d_solver=exponax_kernel=gaussian_correlation_length=0.03_bc=periodic_nu=0.0005_t=1.0_seed=3_N=5_test.pkl"

# 加载 pt
pt_obj = torch.load(pt_file, map_location="cpu", weights_only=False)
x, y, t_final = pt_obj["x"], pt_obj["y"], pt_obj["t_final"]

# 取前 50 条
x50, y50 = x[:10], y[:10]

# 构造和 pkl 一致的 list[dict]
records = []
for i in range(10):
    rec = {
        "index": i,
        "a": x50[i].numpy(),     # numpy.ndarray
        "g(a)": y50[i],          # 保持 tensor/ArrayImpl
        "G(a)": y50[i].numpy()   # numpy.ndarray
    }
    records.append(rec)

# 输出文件名：和 test 文件一致，只是换成 train
new_pkl_file = pkl_file.replace("test", "train")

with open(new_pkl_file, "wb") as f:
    pickle.dump(records, f)

print("保存成功:", new_pkl_file)
```
