```python
from pathlib import Path

folder = "../../../../../datasets/exponax_datasets/t20/generalizability"     # <- change this
files = sorted([p.name for p in Path(folder).iterdir() if p.is_file()])
import re

def parse_family_kernel(fname: str):
    # 匹配示例：
    # ..._GRF_matern_...
    # ..._LogGRF_rbf_...
    # ..._NegLogGRF_periodic_...
    m = re.search(r'(NegLogGRF|LogGRF|GRF)_([A-Za-z0-9]+)', fname)
    if not m:
        return (None, None)
    family = m.group(1)          # 保持原大小写
    kernel = m.group(2).lower()  # 统一小写，避免大小写导致“不同”
    return (family, kernel)

prev_fk = None
for i, file in enumerate(files):
    fk = parse_family_kernel(file)   # (family, kernel)
    if i == 0 or fk != prev_fk:
        print(file)
    prev_fk = fk
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib import colors as mcolors

CSV_PATH = "SUMMARY__Tin10_Tout10.csv"

# ---------- 读取 & 预处理 ----------
df = pd.read_csv(CSV_PATH)
for col in df.columns:
    if col in ["dataset_name", "dataset_path", "type", "solver"]:
        continue
    df[col] = pd.to_numeric(df[col], errors="ignore")

# ---- 解析 (family, kernel) ----
def extract_family_kernel(row):
    txts = []
    for key in ("type", "dataset_name"):
        v = row.get(key, None)
        if isinstance(v, str):
            txts.append(v)
    txt = " ".join(txts).lower()

    # family（与 kernel 独立）
    if "negloggrf" in txt:
        family = "NegLogGRF"
    elif "loggrf" in txt:
        family = "LogGRF"
    elif "grf" in txt:
        family = "GRF"
    else:
        family = "other"

    # kernel（多路兜底）
    kernel_raw = None
    m = re.search(r"kernel\s*=\s*([a-z0-9_]+)", txt)
    if m:
        kernel_raw = m.group(1)
    if not kernel_raw:
        m = re.search(r"(?:negloggrf|loggrf|grf)[\-_]?([a-z0-9]+)", txt)
        if m:
            kernel_raw = m.group(1)
    if not kernel_raw:
        for kw in [
            "matern", "periodic", "gaussian", "rbf", "rq",
            "rational_quadratic","rationalquadratic",
            "se","sqexp","squared_exponential","squaredexponential",
            "gauss","exp_quadratic","expquadratic"
        ]:
            if kw in txt:
                kernel_raw = kw
                break

    def canon(k):
        if not k:
            return "other"
        k = k.lower()
        if k in {"gauss","gaussian","rbf","se","sqexp","squared_exponential",
                 "squaredexponential","exp_quadratic","expquadratic"}:
            return "gaussian"
        if k in {"rq","rational_quadratic","rationalquadratic"}:
            return "rq"
        return k

    kernel = canon(kernel_raw)
    return pd.Series({"family": family, "kernel": kernel})

df[["family", "kernel"]] = df.apply(extract_family_kernel, axis=1)

# ---- 生成 range 标签（根据 vmin/vmax）----
def fmt_float(v):
    if pd.isna(v):
        return "?"
    try:
        return f"{float(v):.6g}"
    except Exception:
        return str(v)

def build_range_label(row):
    vmin = row.get("vmin", np.nan)
    vmax = row.get("vmax", np.nan)
    if pd.isna(vmin) or pd.isna(vmax):
        return "range=unknown"
    lo, hi = float(vmin), float(vmax)
    if lo > hi:
        lo, hi = hi, lo
    return f"range=({fmt_float(lo)},{fmt_float(hi)})"

df["range_label"] = df.apply(build_range_label, axis=1)

# ---------- 只保留三大族 ----------
families = ["GRF", "LogGRF", "NegLogGRF"]
df = df[df["family"].isin(families)].copy()

# ---------- 颜色：range -> 色相 ----------
all_ranges = df["range_label"].dropna().unique().tolist()
palette = plt.cm.Set2.colors if len(all_ranges) <= 8 else plt.cm.tab20.colors
range2color = {rg: palette[i % len(palette)] for i, rg in enumerate(all_ranges)}
def row_color(row):
    return range2color.get(row["range_label"], (0.7, 0.7, 0.7))

# ---------- 纹理：kernel -> hatch ----------
hatch_map = {
    "gaussian": "-",
    "rbf": ".",
    "rq": "\\",
    "matern": "/",
    "periodic": "x",
    "other": None,
}

# ---------- 统一排序（保证三图一致） ----------
kernel_order = ["gaussian", "rbf", "rq", "matern", "periodic", "other"]
range_order  = all_ranges

def sort_key(row):
    k = row["kernel"]
    r = row["range_label"]
    ko = kernel_order.index(k) if k in kernel_order else len(kernel_order)
    ro = range_order.index(r) if r in range_order else len(range_order)
    rmse = float(row["rmse_mean"]) if pd.notna(row["rmse_mean"]) else np.inf
    return (ko, ro, rmse)

df_sorted = (
    df.assign(__sort=df.apply(sort_key, axis=1))
      .sort_values(by="__sort", kind="mergesort")
      .drop(columns="__sort")
      .reset_index(drop=True)
)

# ---------- 统一 y 轴范围（log） ----------
rmse_vals = df_sorted["rmse_mean"].astype(float).replace([np.inf, -np.inf], np.nan)
rmse_vals = rmse_vals[(rmse_vals > 0)].dropna()
ymin_plot = (rmse_vals.min() * 0.8) if len(rmse_vals) else 1e-3
ymax_plot = (rmse_vals.max() * 1.2) if len(rmse_vals) else 1.0

# ---------- 工具：计算某个子集里各 kernel 的中点刻度 + 分界线 ----------
def kernel_ticks_and_boundaries(sub_df):
    """返回 (tick_pos, tick_lab, boundaries) 基于 sub_df 的当前顺序"""
    tick_pos, tick_lab, boundaries = [], [], []
    last_end = None
    arr = sub_df["kernel"].values
    for k in kernel_order:
        idx = np.where(arr == k)[0]
        if idx.size == 0:
            continue
        start, end = int(idx[0]), int(idx[-1])
        tick_pos.append((start + end) / 2.0)
        tick_lab.append(k)
        if last_end is not None:
            boundaries.append(last_end + 0.5)
        last_end = end
    return tick_pos, tick_lab, boundaries

# ---------- 逐 family 作图（只 RMSE） ----------
subsets = [df_sorted[df_sorted["family"] == fam].reset_index(drop=True) for fam in families]
n_bars_max = max((len(s) for s in subsets), default=0)
fig_w = max(20, int(n_bars_max * 0.06))

fig, axes = plt.subplots(3, 1, figsize=(fig_w, 12), sharey=True)
if not isinstance(axes, np.ndarray):
    axes = [axes]

for ax, fam, sub in zip(axes, families, subsets):
    x = np.arange(len(sub))
    heights = sub["rmse_mean"].values.astype(float)

    bars = ax.bar(
        x, heights,
        color=[row_color(r) for _, r in sub.iterrows()],
        edgecolor="black", linewidth=0.0  # 若你坚持“无边框”，可改 edgecolor="none"（但 hatch 会变不明显）
    )
    for b, (_, r) in zip(bars, sub.iterrows()):
        h = hatch_map.get(r["kernel"], None)
        if h:
            b.set_hatch(h)

    ax.set_yscale("log")
    ax.set_ylim(ymin_plot, ymax_plot)
    ax.set_axisbelow(True)  # 让网格画在柱子下方
    ax.grid(axis='y', which='major', linestyle='--', linewidth=3.6, alpha=0.7)
    ax.grid(axis='y', which='minor', linestyle=':',  linewidth=2, alpha=0.5)
    ax.set_xlim(-0.5, len(sub) - 0.5 if len(sub) else 0.5)
    ax.set_title(f"{fam} (n={len(sub)}) — RMSE", fontsize=18)
    ax.set_ylabel("RMSE (log)", fontsize=15)

    # —— 核心：x 轴稀疏标注 & 分界线（按 kernel 块）——
    tick_pos, tick_lab, boundaries = kernel_ticks_and_boundaries(sub)
    # 只在底部 panel 显示标签；上两行仅显示分界线
    if fam == families[-1]:
        ax.set_xticks(tick_pos)
        ax.set_xticklabels(tick_lab, rotation=0, fontsize=15)
    else:
        ax.set_xticks([])
    for b in boundaries:
        ax.axvline(b, color="gray", lw=0.6, ls="--", alpha=0.4)

# ---------- 图例 ----------
legend_elements = []
# 1) range（色相）
for rg in all_ranges:
    legend_elements.append(Patch(facecolor=range2color[rg], edgecolor="none", label=rg))
# 2) kernel（hatch）
for k, h in hatch_map.items():
    if not h:
        continue
    legend_elements.append(Patch(facecolor="white", edgecolor="black", hatch=h, label=f"kernel={k}", linewidth=0.0))

fig.legend(handles=legend_elements, bbox_to_anchor=(1.01, 0.5), loc="center left", title="Legend")
fig.tight_layout(rect=[0, 0, 0.95, 1])
plt.show()
```
