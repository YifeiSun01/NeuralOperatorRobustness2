```python
import torch
import matplotlib.pyplot as plt

# 假设你的数据是这样：
dataset = torch.concat([data["x_adv"],data["y"]],axis=-1)  # shape [1, H, W, T]

# 取出 batch 维
frames = dataset[0]  # shape [256, 256, 20]

# 创建画布
fig, axs = plt.subplots(5, 5, figsize=(15, 15))  # 4行5列

for i in range(21):
    ax = axs[i // 5, i % 5]   # 行列索引
    im = ax.imshow(frames[:, :, i].cpu(), cmap="viridis")
    ax.set_title(f"Frame {i+1}")
    ax.axis("off")

# 统一加一个 colorbar（如果你想每个子图单独一个也行）
fig.colorbar(im, ax=axs, orientation="vertical", fraction=0.02, pad=0.04)

plt.tight_layout()
plt.show()
```
