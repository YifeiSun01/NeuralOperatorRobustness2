```python
import inspect
import exponax as ex

nsmod = ex.stepper._navier_stokes

print("=== 所有可见成员 ===")
cands = [n for n in dir(nsmod) if not n.startswith("_")]
print(cands)

print("\n=== 逐个打印签名与文档（若有）===")
for name in cands:
    obj = getattr(nsmod, name)
    if callable(obj):
        print(f"\n--- {name} ---")
        try:
            sig = inspect.signature(obj)
            print("Signature:", sig)
        except Exception as e:
            print("Signature: <unavailable>", e)
        try:
            doc = inspect.getdoc(obj)
            print("Doc:\n", doc if doc else "<no docstring>")
        except Exception as e:
            print("Doc: <unavailable>", e)
```
