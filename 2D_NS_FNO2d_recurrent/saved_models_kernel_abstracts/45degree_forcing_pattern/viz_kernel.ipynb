```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

def _to_numpy(x: torch.Tensor) -> np.ndarray:
    return x.detach().cpu().numpy()

def _data_log_transform(arr: np.ndarray, mode: str):
    """
    数据级对数变换:
      - 'none': 不变换
      - 'log': 仅适用于 arr>0
      - 'log1p': 对 arr 做 log(1+x)，要求 arr>-1
      - 'signed_log1p': sign(x)*log(1+|x|)，可处理正负
    """
    if mode == "none":
        return arr
    if mode == "log":
        return np.log(arr)
    if mode == "log1p":
        return np.log1p(arr)
    if mode == "signed_log1p":
        return np.sign(arr) * np.log1p(np.abs(arr))
    raise ValueError(f"Unknown mode: {mode}")

def plot_spatial_heatmaps(
    data,
    layer="0",
    top_singval_k=4,
    save_dir=None,
    *,
    data_log_mode="none",      # 'none' | 'log' | 'log1p' | 'signed_log1p'
    color_log_scale=False,     # True -> 颜色轴对数(LogNorm)
    color_log_eps=None,        # 若需要，可加偏移使数据为正：v -> v + eps
    cmap=None,                 # 可传入自定义 colormap
    s2_eps=1e-12               # 计算 σ1/σ2-1 时 σ2 的下限，防止除零
):
    """
    - data_log_mode: 逐 cell 数据级对数变换（真的把数值变了）
    - color_log_scale: 颜色映射使用对数刻度（不改数据）
    - color_log_eps: 若 color_log_scale=True 且数据可能<=0，可加一个全局偏移确保 >0
    """
    spatial = data["layers"][layer]["spatial"]
    spatial_extra = data["layers"][layer].get("spatial_extra", {})

    def _imshow(ax, arr):
        # 可选：为 LogNorm 做正数性保障
        arr_to_plot = arr
        if color_log_scale:
            plot_vals = arr_to_plot
            if color_log_eps is not None:
                plot_vals = plot_vals + color_log_eps
            vmin = np.nanmin(plot_vals)
            if vmin <= 0:
                raise ValueError(
                    "color_log_scale=True 但存在非正值；"
                    "请先选择 data_log_mode 使数据为正，或提供 color_log_eps> -min(data)。"
                )
            norm = mcolors.LogNorm(vmin=np.nanmin(plot_vals), vmax=np.nanmax(plot_vals))
            im = ax.imshow(plot_vals, origin="lower", norm=norm, cmap=cmap)
        else:
            im = ax.imshow(arr_to_plot, origin="lower", cmap=cmap)
        return im

    # ----- 2D fields -----
    for group_name, group in [("spatial", spatial), ("spatial_extra", spatial_extra)]:
        for key, tensor in group.items():
            arr = _to_numpy(tensor)
            if arr.ndim != 2:
                continue
            # 数据级对数变换（如需）
            arr = _data_log_transform(arr, data_log_mode)
            fig, ax = plt.subplots()
            im = _imshow(ax, arr)
            plt.colorbar(im, ax=ax)
            ax.set_title(f"{group_name}.{key} [{data_log_mode}{' + LogNorm' if color_log_scale else ''}]")
            fig.tight_layout()
            if save_dir:
                fig.savefig(f"{save_dir}/{group_name}.{key}.png", dpi=150)
            plt.show()

    # ----- 3D TopSingVals (H, W, K) -----
    if "TopSingVals" in spatial:
        tsv = _to_numpy(spatial["TopSingVals"])
        H, W, K = tsv.shape

        # (A) 逐切片作图（前 k 个）
        k_use = min(top_singval_k, K)
        for i in range(k_use):
            arr = _data_log_transform(tsv[..., i], data_log_mode)
            fig, ax = plt.subplots()
            im = _imshow(ax, arr)
            plt.colorbar(im, ax=ax)
            ax.set_title(f"spatial.TopSingVals[..., {i}] [{data_log_mode}{' + LogNorm' if color_log_scale else ''}]")
            fig.tight_layout()
            if save_dir:
                fig.savefig(f"{save_dir}/spatial.TopSingVals_{i}.png", dpi=150)
            plt.show()

        # (B) 新增：最大奇异值 / 第二大奇异值 − 1 的图
        if K >= 2:
            s1 = tsv[..., 0]
            s2 = tsv[..., 1]
            ratio_minus1 = (s1 / np.clip(s2, s2_eps, None)) - 1.0  # σ1/σ2 - 1

            arr = _data_log_transform(ratio_minus1, data_log_mode)
            fig, ax = plt.subplots()
            im = _imshow(ax, arr)
            plt.colorbar(im, ax=ax)
            ax.set_title(f"spatial.S1_over_S2_minus1 [{data_log_mode}{' + LogNorm' if color_log_scale else ''}]")
            fig.tight_layout()
            if save_dir:
                fig.savefig(f"{save_dir}/spatial.S1_over_S2_minus1.png", dpi=150)
            plt.show()
        else:
            print("[warn] TopSingVals 的最后一维 < 2，无法计算 σ1/σ2 - 1")


# ---- usage ----
plot_spatial_heatmaps(data, layer="0", top_singval_k=4, save_dir=None)
plot_spatial_heatmaps(data, layer="0", top_singval_k=4, save_dir=None, data_log_mode="log")
```

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors  # Normalize / LogNorm

def _tonp(x: torch.Tensor) -> np.ndarray:
    return x.detach().cpu().numpy()

def _data_log_transform(arr: np.ndarray, mode: str) -> np.ndarray:
    """
    数据级对数变换:
      - 'none': 不变换
      - 'log':    np.log(x)        (要求 x > 0)
      - 'log1p':  np.log1p(x)      (常用于含0或小正数; x > -1)
      - 'signed_log1p': sign(x)*log1p(|x|)  (保留符号, 压缩动态范围)
    """
    if mode == "none":
        return arr
    if mode == "log":
        return np.log(arr)
    if mode == "log1p":
        return np.log1p(arr)       # 官方文档说明 log(1+x) 的数值稳定性
    if mode == "signed_log1p":
        return np.sign(arr) * np.log1p(np.abs(arr))
    raise ValueError(f"Unknown data_log_mode: {mode}")

def _pair_norm_linear(A: np.ndarray, B: np.ndarray, symmetric=False):
    if symmetric:
        m = max(np.nanmax(np.abs(A)), np.nanmax(np.abs(B)))
        return mcolors.Normalize(vmin=-m, vmax=+m)   # 对称色标（线性）
    vmin = min(np.nanmin(A), np.nanmin(B))
    vmax = max(np.nanmax(A), np.nanmax(B))
    return mcolors.Normalize(vmin=vmin, vmax=vmax)

def _pair_norm_log(A: np.ndarray, B: np.ndarray, eps: float | None):
    """
    颜色轴对数刻度的共享 LogNorm。
    要求 A,B 全正；若不保证，则用 eps 平移：X -> X + eps，并确保 min>0。
    """
    if eps is not None:
        A = A + eps
        B = B + eps
    vmin = min(np.nanmin(A), np.nanmin(B))
    vmax = max(np.nanmax(A), np.nanmax(B))
    if not (np.isfinite(vmin) and np.isfinite(vmax)) or vmin <= 0:
        raise ValueError("color_log_scale=True 但存在非正或非有限值；请设置 color_log_eps 或使用数据级对数。")
    return mcolors.LogNorm(vmin=vmin, vmax=vmax), A, B  # LogNorm 需正值

def plot_freq_pair(
    data,
    layer: str = "0",
    tsv_indices=(0,1,2),
    symmetric: bool = False,
    save_dir: str | None = None,
    *,
    data_log_mode: str = "none",     # 'none' | 'log' | 'log1p' | 'signed_log1p'
    color_log_scale: bool = False,   # 对颜色轴使用 LogNorm（数据需全正或加偏移）
    color_log_eps: float | None = None,
    cmap=None,
    s2_eps: float = 1e-12            # 计算 σ1/σ2-1 时 σ2 的下限，防止除零
):
    """
    将 freq_neg 与 freq_pos 的同名指标并排显示，使用共享色标（线性或对数）。
    - 若 color_log_scale=True -> 使用 LogNorm（不改数据；必要时用 color_log_eps 平移保证 >0）。
    - 若 data_log_mode != 'none' -> 对 A,B 逐 cell 做变换，再共享线性色标（或再叠加 LogNorm 也行，但通常二者择一）。
    参考：imshow / origin='lower'、LogNorm、colorbar 的官方文档。
    """
    fneg = data["layers"][layer]["freq_neg"]
    fpos = data["layers"][layer]["freq_pos"]

    common_keys = [k for k in fneg.keys() if k in fpos and k != "TopSingVals"]

    # --- (1) 标准 2D 指标并排 ---
    for k in common_keys:
        A = _tonp(fneg[k]); B = _tonp(fpos[k])
        if A.ndim != 2 or B.ndim != 2:
            continue

        # 数据级对数（如需）
        A_t = _data_log_transform(A, data_log_mode)
        B_t = _data_log_transform(B, data_log_mode)

        # 颜色轴归一化
        if color_log_scale:
            norm, A_plot, B_plot = _pair_norm_log(A_t, B_t, color_log_eps)
        else:
            norm = _pair_norm_linear(A_t, B_t, symmetric=symmetric)
            A_plot, B_plot = A_t, B_t

        fig, axs = plt.subplots(1, 2, figsize=(8, 3.6), layout="constrained")
        im0 = axs[0].imshow(A_plot, origin="lower", norm=norm, cmap=cmap)  # imshow + 共享norm
        axs[0].set_title(f"freq_neg.{k}")
        im1 = axs[1].imshow(B_plot, origin="lower", norm=norm, cmap=cmap)
        axs[1].set_title(f"freq_pos.{k}")
        fig.colorbar(im1, ax=axs, shrink=0.85)  # 单个 colorbar 表示两图的共享色标

        if save_dir:
            fig.savefig(f"{save_dir}/freq_pair_{k}.png", dpi=150)
        plt.show()

    # --- (2) TopSingVals: 逐切片并排 ---
    if "TopSingVals" in fneg and "TopSingVals" in fpos:
        A3 = _tonp(fneg["TopSingVals"])
        B3 = _tonp(fpos["TopSingVals"])
        H, W, K = A3.shape
        for idx in tsv_indices:
            if idx < 0 or idx >= K:
                continue
            A = A3[..., idx]; B = B3[..., idx]

            # 数据级对数
            A_t = _data_log_transform(A, data_log_mode)
            B_t = _data_log_transform(B, data_log_mode)

            # 颜色轴
            if color_log_scale:
                norm, A_plot, B_plot = _pair_norm_log(A_t, B_t, color_log_eps)
            else:
                norm = _pair_norm_linear(A_t, B_t, symmetric=symmetric)
                A_plot, B_plot = A_t, B_t

            fig, axs = plt.subplots(1, 2, figsize=(8, 3.6), layout="constrained")
            im0 = axs[0].imshow(A_plot, origin="lower", norm=norm, cmap=cmap)
            axs[0].set_title(f"freq_neg.TopSingVals[..., {idx}]")
            im1 = axs[1].imshow(B_plot, origin="lower", norm=norm, cmap=cmap)
            axs[1].set_title(f"freq_pos.TopSingVals[..., {idx}]")
            fig.colorbar(im1, ax=axs, shrink=0.85)

            if save_dir:
                fig.savefig(f"{save_dir}/freq_pair_TopSingVals_{idx}.png", dpi=150)
            plt.show()

        # --- (3) 新增：σ1/σ2 - 1 成对并排图 ---
        if K >= 2:
            s1_neg, s2_neg = A3[..., 0], A3[..., 1]
            s1_pos, s2_pos = B3[..., 0], B3[..., 1]
            R_neg = (s1_neg / np.clip(s2_neg, s2_eps, None)) - 1.0   # σ1/σ2 - 1
            R_pos = (s1_pos / np.clip(s2_pos, s2_eps, None)) - 1.0

            # 数据级对数（如需）
            Rn_t = _data_log_transform(R_neg, data_log_mode)
            Rp_t = _data_log_transform(R_pos, data_log_mode)

            # 颜色轴（共享范围）
            if color_log_scale:
                norm, Rn_plot, Rp_plot = _pair_norm_log(Rn_t, Rp_t, color_log_eps)
            else:
                norm = _pair_norm_linear(Rn_t, Rp_t, symmetric=symmetric)
                Rn_plot, Rp_plot = Rn_t, Rp_t

            fig, axs = plt.subplots(1, 2, figsize=(8, 3.6), layout="constrained")
            im0 = axs[0].imshow(Rn_plot, origin="lower", norm=norm, cmap=cmap)
            axs[0].set_title(f"freq_neg.S1_over_S2_minus1")
            im1 = axs[1].imshow(Rp_plot, origin="lower", norm=norm, cmap=cmap)
            axs[1].set_title(f"freq_pos.S1_over_S2_minus1")
            fig.colorbar(im1, ax=axs, shrink=0.85)

            if save_dir:
                fig.savefig(f"{save_dir}/freq_pair_S1_over_S2_minus1.png", dpi=150)
            plt.show()
        else:
            print("[warn] TopSingVals 的最后一维 < 2，无法计算 σ1/σ2 - 1")


# Example:
plot_freq_pair(data, layer="0", tsv_indices=(0,1,2,3), symmetric=False, save_dir=None)
plot_freq_pair(data, layer="0", tsv_indices=(0,1,2,3), symmetric=False, save_dir=None, data_log_mode="log")
```
