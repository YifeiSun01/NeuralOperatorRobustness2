```python
from pathlib import Path
import os

def print_tree(
    root=".",
    max_depth=None,
    ignore_hidden=True,
    ignore_names=None,   # e.g. {"__pycache__", ".git", ".ipynb_checkpoints"}
    dirs_first=True,
):
    """
    Pretty-print a directory tree (like the `tree` CLI) in a Jupyter notebook.

    Args:
        root (str|Path): directory to list
        max_depth (int|None): limit depth; None = unlimited
        ignore_hidden (bool): skip names starting with '.'
        ignore_names (set[str]|None): skip these exact names
        dirs_first (bool): list directories before files
    """
    root = Path(root).expanduser().resolve()
    ignore_names = set(ignore_names or [])

    def _listdir(p: Path):
        try:
            items = [e for e in p.iterdir()
                     if (not ignore_hidden or not e.name.startswith("."))
                     and e.name not in ignore_names]
        except PermissionError:
            return []
        if dirs_first:
            items.sort(key=lambda e: (not e.is_dir(), e.name.lower()))
        else:
            items.sort(key=lambda e: e.name.lower())
        return items

    def _recurse(p: Path, prefix="", depth=0):
        children = _listdir(p)
        total = len(children)
        for i, child in enumerate(children):
            connector = "└── " if i == total - 1 else "├── "
            print(prefix + connector + child.name)
            if child.is_dir():
                if max_depth is None or depth + 1 < max_depth:
                    extension = "    " if i == total - 1 else "│   "
                    _recurse(child, prefix + extension, depth + 1)

    # header
    print(str(root))
    _recurse(root)
```

```python
print_tree("2D", ignore_names={"__pycache__", ".git", ".ipynb_checkpoints"})
```
