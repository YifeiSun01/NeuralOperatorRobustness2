```python
import os
from pathlib import Path

def print_file_tree(root: str, prefix: str = ""):
    """
    打印文件夹下的文件树（递归展开）
    root   : 要查看的文件夹路径
    prefix : 内部参数，用来控制缩进（调用时不用管）
    """
    root_path = Path(root)
    if not root_path.exists():
        print(f"[ERROR] 路径不存在: {root}")
        return
    
    entries = sorted(root_path.iterdir(), key=lambda x: (x.is_file(), x.name.lower()))
    count = len(entries)
    
    for i, entry in enumerate(entries):
        connector = "└── " if i == count - 1 else "├── "
        print(prefix + connector + entry.name)
        
        if entry.is_dir():
            extension = "    " if i == count - 1 else "│   "
            print_file_tree(entry, prefix + extension)

# 在 notebook 里调用示例：
print_file_tree("../saved_models_expanded")
```
