```python
import pandas as pd

# ===== You defined these two groups =====
DATASET_COLS = [
    'type','solver','nx','N','nu','t','ntimepoints','length_scale','variance','alpha',
    'tau','period','exp_factor','vmax','vmin','dataset_group','dataset_name',
    'dataset_path','n'
]
MODEL_COLS = [
    'model_group','model_name','model_path','model_source_dir','modes1','modes2',
    'width','Tin','Tout','step','epochs','seed','strategy','pct_tag','percent',
    'expanded_stem','expanded_file_name','expanded_file_path','orig_train_dataset',
    'with_replacement','k_requested','N_in_file'
]

def _levels_with_variation(df: pd.DataFrame, cols):
    """Keep only columns that exist and have >=2 non-NaN unique values (i.e., provide discrimination)."""
    present = [c for c in cols if c in df.columns]
    keep = []
    for c in present:
        if df[c].dropna().nunique() >= 2:
            keep.append(c)
    return keep

def build_pivot_keep_na(df: pd.DataFrame, value='rmse_mean', aggfunc='mean'):
    # 1) auto-pick discriminative levels
    ds_levels = _levels_with_variation(df, DATASET_COLS)
    model_levels = _levels_with_variation(df, MODEL_COLS)

    # 2) drop ultra–high-cardinality path/name keys from levels (prevents explosion)
    drop_row_keys = {'dataset_path'}
    drop_col_keys = {'model_path','model_source_dir',
                     'expanded_stem','expanded_file_name','expanded_file_path'}
    ds_levels = [c for c in ds_levels if c not in drop_row_keys]
    model_levels = [c for c in model_levels if c not in drop_col_keys]

    # 3) DO NOT FILTER NaN rows — instead, fill NaN in *keys only* with a sentinel so nothing is dropped
    SENTINEL = '__NA__'
    df2 = df.copy()
    key_cols = list(set(ds_levels + model_levels))
    for c in key_cols:
        df2[c] = df2[c].astype('object').where(df2[c].notna(), SENTINEL)

    # 4) ensure there is at least one level on each axis
    if not ds_levels:
        const_row = '__ALL_DATASETS__'
        df2[const_row] = const_row
        ds_levels = [const_row]
    if not model_levels:
        const_col = '__ALL_MODELS__'
        df2[const_col] = const_col
        model_levels = [const_col]

    # 5) pivot (no rows dropped due to NaN because of sentinel; only value is aggregated)
    pivot = pd.pivot_table(
        df2,
        index=ds_levels,
        columns=model_levels,
        values=value,
        aggfunc=aggfunc,
        observed=False
    ).sort_index().sort_index(axis=1)

    # 6) friendly names
    if isinstance(pivot.index, pd.MultiIndex):
        pivot.index.names = ds_levels
    else:
        pivot.index.name = ds_levels[0]
    if isinstance(pivot.columns, pd.MultiIndex):
        pivot.columns.names = model_levels
    else:
        pivot.columns.name = model_levels[0]

    return pivot, ds_levels, model_levels

def print_index_level_uniques(pivot: pd.DataFrame):
    """Print all unique values of each level of the *row index* used in the pivot."""
    if isinstance(pivot.index, pd.MultiIndex):
        for lvl, name in enumerate(pivot.index.names):
            vals = pivot.index.get_level_values(lvl).unique().tolist()
            print(f"[index level {lvl} - {name}] ({len(vals)})")
            print(vals)
    else:
        vals = pivot.index.unique().tolist()
        print(f"[index level 0 - {pivot.index.name}] ({len(vals)})")
        print(vals)

def print_column_level_uniques(pivot: pd.DataFrame):
    """(Optional) Also print unique values of each level of the *column index*."""
    if isinstance(pivot.columns, pd.MultiIndex):
        for lvl, name in enumerate(pivot.columns.names):
            vals = pivot.columns.get_level_values(lvl).unique().tolist()
            print(f"[columns level {lvl} - {name}] ({len(vals)})")
            print(vals)
    else:
        vals = pivot.columns.unique().tolist()
        print(f"[columns level 0 - {pivot.columns.name}] ({len(vals)})")
        print(vals)

# ====== Usage ======
# assume your original DataFrame is df
pivot, row_levels, col_levels = build_pivot_keep_na(df, value='rmse_mean', aggfunc='mean')
print("row_levels:", row_levels)
print("col_levels:", col_levels)

# print unique values of each row-index level (as requested)
print_index_level_uniques(pivot)
pivot
```

```python
import pandas as pd

SENTINEL = '__NA__'

DATASET_COLS = [
    'type','solver','nx','N','nu','t','ntimepoints','length_scale','variance','alpha',
    'tau','period','exp_factor','vmax','vmin','dataset_group','dataset_name',
    'dataset_path','n'
]
MODEL_COLS = [
    'model_group','model_name','model_path','model_source_dir','modes1','modes2',
    'width','Tin','Tout','step','epochs','seed','strategy','pct_tag','percent',
    'expanded_stem','expanded_file_name','expanded_file_path','orig_train_dataset',
    'with_replacement','k_requested','N_in_file'
]

def _levels_with_variation(df, cols):
    present = [c for c in cols if c in df.columns]
    return [c for c in present if df[c].dropna().nunique() >= 2]

def _fill_keys_with_sentinel(df, keys):
    out = df.copy()
    for c in keys:
        out[c] = out[c].astype('object').where(out[c].notna(), SENTINEL)
    return out

def build_pivot_rmse_raw(df, value='rmse_mean', aggfunc='mean'):
    # 自动挑层级；不引入高基数路径列
    ds_levels = _levels_with_variation(df, DATASET_COLS)
    ds_levels = [c for c in ds_levels if c != 'dataset_path']
    model_levels = _levels_with_variation(df, MODEL_COLS)
    model_levels = [c for c in model_levels
                    if c not in {'model_path','model_source_dir',
                                 'expanded_stem','expanded_file_name','expanded_file_path'}]
    # 不丢行：键填占位
    key_cols = list(set(ds_levels + model_levels))
    df2 = _fill_keys_with_sentinel(df, key_cols)

    # 至少保证两侧有层级
    if not ds_levels:
        ds_levels = ['__ALL_DATASETS__']; df2['__ALL_DATASETS__'] = '__ALL_DATASETS__'
    if not model_levels:
        model_levels = ['__ALL_MODELS__']; df2['__ALL_MODELS__'] = '__ALL_MODELS__'

    piv = pd.pivot_table(
        df2, index=ds_levels, columns=model_levels, values=value,
        aggfunc=aggfunc, observed=False
    ).sort_index().sort_index(axis=1)

    # 命名
    if isinstance(piv.index, pd.MultiIndex): piv.index.names = ds_levels
    else: piv.index.name = ds_levels[0]
    if isinstance(piv.columns, pd.MultiIndex): piv.columns.names = model_levels
    else: piv.columns.name = model_levels[0]
    return piv

def subtract_baseline_per_row(pivot: pd.DataFrame, reduce='mean'):
    """
    在 pivot 上按行减 baseline：
    1) 取列层级 model_group=='baseline' 的全部列子集；
    2) 对其按行做 reduce（默认 mean，也可 'first'）得到每行一个 baseline 值；
    3) 整表逐行相减。
    """
    if 'model_group' not in (pivot.columns.names or []):
        raise ValueError("列层级里没有 'model_group'，无法识别 baseline。")

    # 只取 baseline 列子框
    base_block = pivot.xs('baseline', axis=1, level='model_group', drop_level=False)

    # 行方向压缩成一个标量 baseline
    if reduce == 'mean':
        base_series = base_block.mean(axis=1, skipna=True)
    elif reduce == 'first':
        # 每行取第一个非空 baseline 值
        base_series = base_block.apply(lambda r: r.dropna().iloc[0] if r.notna().any() else pd.NA, axis=1)
    else:
        raise ValueError("reduce 仅支持 'mean' 或 'first'")

    # 逐行相减；baseline 缺失则结果为 NaN（保持原样）
    pivot_delta = pivot.subtract(base_series, axis=0)

    return pivot_delta, base_series

# ================== 用法 ==================
# 1) 先做“原始 rmse”的 pivot（你现在看到的那个就是这种形态）
pivot_rmse = build_pivot_rmse_raw(df, value='rmse_mean', aggfunc='mean')

# 2) 在 pivot 上逐行减 baseline
pivot_minus, baseline_series = subtract_baseline_per_row(pivot_rmse, reduce='mean')

# 3) 快速自检：baseline 列应当全为 0（或非常接近 0）
if 'baseline' in pivot_minus.columns.get_level_values('model_group'):
    baseline_block = pivot_minus.xs('baseline', axis=1, level='model_group', drop_level=False)
    print("max|baseline columns| after subtraction =",
          float(baseline_block.abs().max().max()))
else:
    print("warning: 没找到 baseline 列；请确认 'model_group' 层级。")

pivot_minus
pivot_minus.to_csv("pivot_rmse_minus_baseline_from_pivot.csv")
```

```python
# -*- coding: utf-8 -*-
import os
import re
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# ============ 可调参数 ============
SAVE_DIR = "plots/models_rmse_minus"
FIG_DPI = 160
MAX_FIG_W = 80         # 避免超宽
MIN_FIG_W = 12
PX_PER_BAR = 0.05      # 每个 bar 对应的宽度（越大图越宽）

GROUP_ORDER = ["generalizability", "test", "expanded", "train", "other"]
GROUP_COLOR = {
    "generalizability": "#1f77b4",  # 蓝
    "test":              "#ff7f0e",  # 橙
    "expanded":          "#9467bd",  # 紫
    "train":             "#2ca02c",  # 绿
    "other":             "#7f7f7f",  # 灰
}
HATCH_BETTER = "/"  # rmse_minus < 0
HATCH_WORSE  = "."  # rmse_minus >= 0

# ============ 工具函数 ============

def sanitize_filename(s: str) -> str:
    s = re.sub(r"[^\w\-.]+", "_", str(s))
    return s[:230]

def infer_dataset_group_from_index(idx_names, idx_values_row: tuple) -> str:
    """
    从行 MultiIndex 推断 dataset_group。
    1) 若存在 'dataset_group' 层级，直接用；
    2) 否则从 'dataset_name' 或字符串里解析 train/test/expanded/generalizability；
    3) 否则标记为 'other'。
    """
    name2pos = {n: i for i, n in enumerate(idx_names)}
    # 1) 直接取
    if "dataset_group" in name2pos:
        v = idx_values_row[name2pos["dataset_group"]]
        return str(v) if pd.notna(v) else "other"

    # 2) 尝试从 dataset_name 解析
    cand = None
    for key in ("dataset_name",):
        if key in name2pos:
            cand = str(idx_values_row[name2pos[key]]).lower()
            break

    if cand:
        if "generalizability" in cand:
            return "generalizability"
        if "expanded" in cand:
            return "expanded"
        if "_train_" in cand or re.search(r"(?:^|_)train(?:_|$)", cand):
            return "train"
        if "_test_" in cand or re.search(r"(?:^|_)test(?:_|$)", cand):
            return "test"

    return "other"

def subtract_baseline_per_row(pivot: pd.DataFrame, reduce="mean"):
    """
    在 pivot 上按行减 baseline:
    - baseline 通过列层级 'model_group'=='baseline' 识别；
    - 如果没有该层级，原样返回（不做相减）。
    """
    if not isinstance(pivot.columns, pd.MultiIndex):
        return pivot, None  # 单层列，无从识别 baseline

    if "model_group" not in pivot.columns.names:
        return pivot, None

    if "baseline" not in pivot.columns.get_level_values("model_group"):
        return pivot, None

    base_block = pivot.xs("baseline", axis=1, level="model_group", drop_level=False)
    if reduce == "mean":
        base_series = base_block.mean(axis=1, skipna=True)
    elif reduce == "first":
        base_series = base_block.apply(lambda r: r.dropna().iloc[0] if r.notna().any() else np.nan, axis=1)
    else:
        raise ValueError("reduce 仅支持 'mean' 或 'first'")

    pivot_minus = pivot.subtract(base_series, axis=0)
    return pivot_minus, base_series

def ensure_rmse_minus(pivot: pd.DataFrame):
    """
    若 pivot 仍是原始 rmse（baseline 列非 0），则做一次“逐行减 baseline”；
    否则直接返回。
    """
    if isinstance(pivot.columns, pd.MultiIndex) and "model_group" in pivot.columns.names:
        if "baseline" in pivot.columns.get_level_values("model_group"):
            # 粗略判断是否“已减过”：baseline 列是否 ≈ 0
            base_block = pivot.xs("baseline", axis=1, level="model_group", drop_level=False)
            max_abs = float(base_block.abs().max().max())
            if max_abs > 1e-9:
                pivot_minus, _ = subtract_baseline_per_row(pivot, reduce="mean")
                return pivot_minus
    # 没有 baseline 这个概念或已是 0，就直接当作 rmse_minus 使用
    return pivot

def compute_group_layout(pivot: pd.DataFrame):
    """
    返回：
      - row_order: 行的绘制顺序索引
      - group_of_row: 与 row_order 对齐的分组名列表
      - boundaries: 组之间分界线的 x 位置
      - tick_pos, tick_lab: 组中心与标签
    """
    if isinstance(pivot.index, pd.MultiIndex):
        idx_names = pivot.index.names
        rows = list(pivot.index)
    else:
        idx_names = [pivot.index.name or "index"]
        rows = [(v,) for v in pivot.index.values]

    groups = [infer_dataset_group_from_index(idx_names, r) for r in rows]

    # 按既定顺序组织
    buckets = {g: [] for g in GROUP_ORDER}
    # 把不在预定义里的也收进 'other'
    for i, g in enumerate(groups):
        if g not in buckets:
            g = "other"
        buckets[g].append(i)

    # 行顺序（组拼接）
    row_order = []
    group_of_row = []
    for g in GROUP_ORDER:
        idxs = buckets.get(g, [])
        row_order.extend(idxs)
        group_of_row.extend([g] * len(idxs))

    # 计算组中心与边界
    tick_pos, tick_lab, boundaries = [], [], []
    start = 0
    for g in GROUP_ORDER:
        idxs = buckets.get(g, [])
        if not idxs:
            continue
        length = len(idxs)
        mid = start + (length - 1) / 2.0
        tick_pos.append(mid)
        tick_lab.append(g)
        start += length
        boundaries.append(start - 0.5)
    if boundaries:
        boundaries = boundaries[:-1]  # 最后一条是末尾，无需画

    return row_order, group_of_row, tick_pos, tick_lab, boundaries

def plot_one_model(series: pd.Series, group_of_row, tick_pos, tick_lab, boundaries, title, save_path):
    """
    series: 与 row_order 对齐的一维值（rmse_minus），可以包含 NaN
    """
    n = len(series)
    fig_w = max(MIN_FIG_W, min(MAX_FIG_W, int(n * PX_PER_BAR) + 8))
    fig, ax = plt.subplots(figsize=(fig_w, 5))

    x = np.arange(n)
    y = series.to_numpy(float)

    # 颜色：按 dataset_group；纹理：按提升/变差
    colors = [GROUP_COLOR.get(g, GROUP_COLOR["other"]) for g in group_of_row]
    hatches = [HATCH_BETTER if (not pd.isna(v) and v < 0) else HATCH_WORSE for v in y]

    bars = ax.bar(x, y, color=colors, edgecolor="black", linewidth=0.0)
    for b, h in zip(bars, hatches):
        if h:
            b.set_hatch(h)

    # 0 线、边界、网格
    ax.axhline(0.0, color="black", lw=1.0, alpha=0.8)
    for b in boundaries:
        ax.axvline(b, color="gray", lw=0.6, ls="--", alpha=0.5)

    # x 轴刻度只放大类
    ax.set_xticks(tick_pos)
    ax.set_xticklabels(tick_lab, fontsize=11)

    # y 轴范围对称一些
    finite = pd.Series(y).dropna().to_numpy()
    if finite.size:
        yabs = np.nanmax(np.abs(finite))
        ymax = float(yabs * 1.15) if yabs > 0 else 1.0
    else:
        ymax = 1.0
    ax.set_ylim(-ymax, ymax)

    ax.set_xlim(-0.5, n - 0.5 if n else 0.5)
    ax.set_title(title, fontsize=13)
    ax.set_ylabel("RMSE minus baseline", fontsize=11)
    ax.grid(axis="y", linestyle="--", alpha=0.5)

    # 图例
    legend_elements = []
    for g in [g for g in GROUP_ORDER if g in group_of_row]:
        legend_elements.append(Patch(facecolor=GROUP_COLOR[g], edgecolor="none", label=g))
    legend_elements.append(Patch(facecolor="white", edgecolor="black", hatch=HATCH_BETTER, label="better (<0)", linewidth=0.0))
    legend_elements.append(Patch(facecolor="white", edgecolor="black", hatch=HATCH_WORSE,  label="worse (≥0)", linewidth=0.0))
    ax.legend(handles=legend_elements, loc="upper left", bbox_to_anchor=(1.01, 1.0), title="Legend")

    plt.tight_layout(rect=[0, 0, 0.95, 1])
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    plt.savefig(save_path, dpi=FIG_DPI, bbox_inches="tight")
    plt.close(fig)

def model_col_iter(pivot_minus: pd.DataFrame):
    """
    迭代所有“非 baseline”的模型列。
    返回 (col_key, col_title, series)：
      - col_key: 用于文件名的 tuple 或标量
      - col_title: 图标题里友好显示
      - series: 该列的值（Index 与 pivot 行一致）
    """
    if isinstance(pivot_minus.columns, pd.MultiIndex):
        # 过滤 baseline
        col_mask = np.ones(len(pivot_minus.columns), dtype=bool)
        if "model_group" in pivot_minus.columns.names:
            col_mask &= (pivot_minus.columns.get_level_values("model_group") != "baseline")

        cols = [c for c, m in zip(pivot_minus.columns, col_mask) if m]

        for c in cols:
            s = pivot_minus[c]
            # 标题友好化
            parts = []
            for name, val in zip(pivot_minus.columns.names, c if isinstance(c, tuple) else (c,)):
                parts.append(f"{name}={val}")
            title = " | ".join(parts)
            yield c, title, s
    else:
        # 单层列：把名为 baseline 的列去掉
        for c in pivot_minus.columns:
            if str(c).lower() == "baseline":
                continue
            yield c, f"model={c}", pivot_minus[c]

def plot_all_models_from_pivot(pivot_df: pd.DataFrame):
    """
    主入口：
      - 输入 pivot（可能是原始 rmse，也可能已是 rmse_minus）
      - 自动确保变成 rmse_minus
      - 对每个非 baseline 模型列作图
    """
    pivot_minus = ensure_rmse_minus(pivot_df)

    # 行分组与布局
    row_order, group_of_row, tick_pos, tick_lab, boundaries = compute_group_layout(pivot_minus)
    # 重新按 row_order 排列
    pivot_minus_ord = pivot_minus.iloc[row_order, :]

    # 逐模型绘图
    for col_key, col_title, series in model_col_iter(pivot_minus_ord):
        # series 已与 row_order 对齐
        # 文件名
        if isinstance(col_key, tuple):
            fname = "model_" + "__".join(f"{sanitize_filename(str(v))}" for v in col_key)
        else:
            fname = "model_" + sanitize_filename(str(col_key))
        save_path = os.path.join(SAVE_DIR, fname + ".png")

        title = f"RMSE minus baseline — {col_title}"
        plot_one_model(series, group_of_row, tick_pos, tick_lab, boundaries, title, save_path)

    print(f"保存完成：{SAVE_DIR}")

# =================== 用法示例 ===================
# 你已有的透视表变量名可能是 pivot_delta / pivot_minus / pivot 等。
# 假设叫 pivot_df：
plot_all_models_from_pivot(pivot_minus)
```

```python
# -*- coding: utf-8 -*-
import os
import re
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.cm import get_cmap

# ============ 可调参数 ============
SAVE_DIR = "plots/models_rmse_minus"
FIG_DPI = 160
MAX_FIG_W = 80          # 避免超宽
MIN_FIG_W = 12
PX_PER_BAR = 0.05       # 每个 bar 对应的宽度（越大图越宽）

GROUP_ORDER = ["generalizability", "test", "train", "expanded", "other"]
# 非 generalizability 的固定色
GROUP_BASE_COLOR = {
    "test":      "#ff7f0e",  # 橙
    "train":     "#2ca02c",  # 绿
    "expanded":  "#9467bd",  # 紫
    "other":     "#7f7f7f",  # 灰
}

# kernel 的纹理（hatch）
KERNEL_ORDER = ["gaussian", "rbf", "rq", "matern", "periodic", "other"]
KERNEL_HATCH = {
    "gaussian": "-",
    "rbf": ".",
    "rq": "\\",
    "matern": "/",
    "periodic": "x",
    "other": None,
}

# ============ 基础工具 ============
def sanitize_filename(s: str) -> str:
    s = re.sub(r"[^\w\-.]+", "_", str(s))
    return s[:230]

def sanitize_title(s: str) -> str:
    return sanitize_filename(s)

def str_or_none(x):
    return None if (x is None or (isinstance(x, float) and np.isnan(x))) else str(x)

def as_float_or_nan(x):
    try:
        f = float(x)
        return f
    except Exception:
        return np.nan

def canonical_range_tuple(lo, hi):
    lo = as_float_or_nan(lo)
    hi = as_float_or_nan(hi)
    if np.isnan(lo) or np.isnan(hi):
        return None
    return (lo, hi) if lo <= hi else (hi, lo)

def range_label_from_tuple(rt):
    if rt is None:
        return "range=unknown"
    lo, hi = rt
    return f"range=({lo:g},{hi:g})"

def parse_kernel_from_text(txt: str):
    txt = (txt or "").lower()
    # 优先识别常见写法
    for kw, canon in [
        ("gaussian", "gaussian"), ("rbf", "rbf"), ("se", "gaussian"),
        ("sqexp", "gaussian"), ("squared_exponential", "gaussian"),
        ("exp_quadratic", "gaussian"), ("expquadratic", "gaussian"),
        ("rq", "rq"), ("rational_quadratic", "rq"), ("rationalquadratic", "rq"),
        ("matern", "matern"), ("periodic", "periodic"),
    ]:
        if kw in txt:
            return canon
    # 兜底：type 里类似 GRF_matern
    m = re.search(r"(?:negloggrf|loggrf|grf)[\-_]?([a-z0-9]+)", txt)
    if m:
        return parse_kernel_from_text(m.group(1))
    return "other"

def extract_kernel_from_row(idx_names, row_tuple):
    # 优先从 'type'
    if "type" in idx_names:
        v = row_tuple[idx_names.index("type")]
        k = parse_kernel_from_text(str_or_none(v))
        if k != "other":
            return k
    # 再从 dataset_name
    if "dataset_name" in idx_names:
        v = row_tuple[idx_names.index("dataset_name")]
        k = parse_kernel_from_text(str_or_none(v))
        if k != "other":
            return k
    return "other"

def extract_range_from_row(idx_names, row_tuple):
    # 优先从 vmin/vmax 层解析
    have_vmin = "vmin" in idx_names
    have_vmax = "vmax" in idx_names
    if have_vmin and have_vmax:
        rt = canonical_range_tuple(row_tuple[idx_names.index("vmin")],
                                   row_tuple[idx_names.index("vmax")])
        if rt is not None:
            return rt
    # 兜底：从 dataset_name 文本解析
    if "dataset_name" in idx_names:
        dn = str_or_none(row_tuple[idx_names.index("dataset_name")]) or ""
        dn_low = dn.lower()
        m1 = re.search(r"vmax([\-0-9\.]+).*?vmin([\-0-9\.]+)", dn_low)
        m2 = re.search(r"vmin([\-0-9\.]+).*?vmax([\-0-9\.]+)", dn_low)
        m = m1 or m2
        if m:
            rt = canonical_range_tuple(m.group(1), m.group(2))
            if rt is not None:
                return rt
    return None  # unknown

def infer_group_from_row(idx_names, row_tuple):
    # 直接从 dataset_group 层
    if "dataset_group" in idx_names:
        v = str_or_none(row_tuple[idx_names.index("dataset_group")])
        if v:
            return v
    # 从名字推断
    cand = ""
    if "dataset_name" in idx_names:
        cand = (str_or_none(row_tuple[idx_names.index("dataset_name")]) or "").lower()
    if "generalizability" in cand:
        return "generalizability"
    if "expanded" in cand:
        return "expanded"
    if "_train_" in cand or re.search(r"(?:^|_)train(?:_|$)", cand):
        return "train"
    if "_test_" in cand or re.search(r"(?:^|_)test(?:_|$)", cand):
        return "test"
    return "other"

# ============ baseline 处理 ============
def subtract_baseline_per_row(pivot: pd.DataFrame, reduce="mean"):
    if not isinstance(pivot.columns, pd.MultiIndex):
        return pivot, None
    if "model_group" not in pivot.columns.names:
        return pivot, None
    if "baseline" not in pivot.columns.get_level_values("model_group"):
        return pivot, None

    base_block = pivot.xs("baseline", axis=1, level="model_group", drop_level=False)
    if reduce == "mean":
        base_series = base_block.mean(axis=1, skipna=True)
    elif reduce == "first":
        base_series = base_block.apply(lambda r: r.dropna().iloc[0] if r.notna().any() else np.nan, axis=1)
    else:
        raise ValueError("reduce 仅支持 'mean' 或 'first'")
    pivot_minus = pivot.subtract(base_series, axis=0)
    return pivot_minus, base_series

def ensure_rmse_minus(pivot: pd.DataFrame):
    if isinstance(pivot.columns, pd.MultiIndex) and "model_group" in pivot.columns.names:
        if "baseline" in pivot.columns.get_level_values("model_group"):
            base_block = pivot.xs("baseline", axis=1, level="model_group", drop_level=False)
            max_abs = float(base_block.abs().max().max())
            if max_abs > 1e-9:
                pivot_minus, _ = subtract_baseline_per_row(pivot, reduce="mean")
                return pivot_minus
    return pivot

# ============ generalizability 的颜色映射（按 range） ============
def build_generalizability_range_palette(pivot: pd.DataFrame, cmap_name="coolwarm"):
    """
    返回：dict(range_label_str -> RGBA color)
    使用蓝→红的 diverging 色带（默认 coolwarm），
    按 (lo, hi) 的元组排序后均匀采样（0.15~0.85，避免两端过亮）。
    """
    if isinstance(pivot.index, pd.MultiIndex):
        idx_names = pivot.index.names
        rows = list(pivot.index)
    else:
        idx_names = [pivot.index.name or "index"]
        rows = [(v,) for v in pivot.index.values]

    # 收集 generalizability 的所有 range（用 tuple 排序，稳定可复现）
    ranges = []
    for r in rows:
        if infer_group_from_row(idx_names, r) != "generalizability":
            continue
        rt = extract_range_from_row(idx_names, r)
        ranges.append(rt)

    uniq_tuples = sorted(set(ranges), key=lambda rt: (math.inf, math.inf) if rt is None else (rt[0], rt[1]))
    if not uniq_tuples:
        return {}

    labels = [range_label_from_tuple(rt) for rt in uniq_tuples]
    cmap   = get_cmap(cmap_name)
    lo, hi = 0.15, 0.85
    n      = len(labels)
    vals   = [0.5] if n == 1 else np.linspace(lo, hi, n)
    return {lab: cmap(v) for lab, v in zip(labels, vals)}

# ============ 行排序与布局 ============
import math
import re

def compute_row_order_and_layout(pivot: pd.DataFrame, range_palette: dict):
    """
    返回：
      row_order, None, tick_pos, tick_lab, boundaries, meta_by_new,
      sub_xtick_info  # 包含 generalizability 内部子块的 ticks/labels/boundaries 以及 segments（用于后续统计）
                      # sub_xtick_info = {
                      #   "ticks": [...], "labels": [...], "boundaries": [...],
                      #   "segments": [{"start":int,"end":int,"kernel":str,"range_label":str}, ...]
                      # }
    """
    if isinstance(pivot.index, pd.MultiIndex):
        idx_names = pivot.index.names
        rows = list(pivot.index)
    else:
        idx_names = [pivot.index.name or "index"]
        rows = [(v,) for v in pivot.index.values]

    meta = []
    for i, r in enumerate(rows):
        g  = infer_group_from_row(idx_names, r)
        k  = extract_kernel_from_row(idx_names, r) if g == "generalizability" else "other"
        rt = extract_range_from_row(idx_names, r)   if g == "generalizability" else None
        rl = range_label_from_tuple(rt)             if g == "generalizability" else None
        meta.append({"i": i, "group": g, "kernel": k, "range_tuple": rt, "range_label": rl})

    def range_key(rt):
        if rt is None:
            return (math.inf, math.inf)
        lo, hi = rt
        return (lo, hi)

    order_buckets = {g: [] for g in GROUP_ORDER}
    for m in meta:
        order_buckets.setdefault(m["group"], [])
        order_buckets[m["group"]].append(m)

    row_order = []
    tick_pos, tick_lab, boundaries = [], [], []
    gen_sub_ticks, gen_sub_labels, gen_sub_boundaries = [], [], []
    gen_segments = []

    start_running = 0

    for gi, g in enumerate(GROUP_ORDER):
        items = order_buckets.get(g, [])
        if not items:
            continue

        if g == "generalizability":
            items = sorted(
                items,
                key=lambda m: (
                    KERNEL_ORDER.index(m["kernel"]) if m["kernel"] in KERNEL_ORDER else len(KERNEL_ORDER),
                    range_key(m["range_tuple"]),
                    m["i"],
                ),
            )
            # —— 计算 generalizability 内部的 (kernel, range) 子块 —— #
            if items:
                cur_key = None
                seg_start_local = None
                for j, m in enumerate(items):
                    key = (m["kernel"], m["range_label"])
                    if cur_key is None:
                        cur_key = key
                        seg_start_local = j
                    elif key != cur_key:
                        # 结束上一段
                        start_index = start_running + seg_start_local
                        end_index   = start_running + j - 1
                        mid = (start_index + end_index) / 2.0

                        # 简化 range 显示
                        rg = cur_key[1] or "range=unknown"
                        rg_short = re.sub(r"^range=", "", rg)

                        gen_sub_ticks.append(mid)
                        gen_sub_labels.append(f"{cur_key[0]}\n{rg_short}")
                        gen_sub_boundaries.append(start_running + j - 0.5)
                        gen_segments.append({
                            "start": start_index,
                            "end": end_index,
                            "kernel": cur_key[0],
                            "range_label": cur_key[1],
                        })
                        # 新段
                        cur_key = key
                        seg_start_local = j

                # 收尾最后一段
                start_index = start_running + seg_start_local
                end_index   = start_running + len(items) - 1
                mid = (start_index + end_index) / 2.0
                rg = cur_key[1] or "range=unknown"
                rg_short = re.sub(r"^range=", "", rg)
                gen_sub_ticks.append(mid)
                gen_sub_labels.append(f"{cur_key[0]}\n{rg_short}")
                gen_segments.append({
                    "start": start_index,
                    "end": end_index,
                    "kernel": cur_key[0],
                    "range_label": cur_key[1],
                })
        else:
            items = sorted(items, key=lambda m: m["i"])

        cnt = len(items)
        mid = start_running + (cnt - 1) / 2.0
        tick_pos.append(mid)
        tick_lab.append(g)

        row_order.extend([m["i"] for m in items])
        start_running += cnt

        if gi < len(GROUP_ORDER) - 1 and start_running > 0:
            boundaries.append(start_running - 0.5)

    meta_by_new = [meta[i] for i in row_order]

    sub_xtick_info = {
        "ticks": gen_sub_ticks,
        "labels": gen_sub_labels,
        "boundaries": gen_sub_boundaries,
        "segments": gen_segments,
    }
    return row_order, None, tick_pos, tick_lab, boundaries, meta_by_new, sub_xtick_info

# ============ 逐模型绘图 ============
def summarize_good_bad(series: pd.Series, meta_by_new: list):
    """
    只统计 generalizability 行：
      - overall (good/bad)
      - by kernel
      - by range_label
    返回三个 dict：overall, by_kernel, by_range
    """
    overall = {"better": 0, "worse": 0}
    by_kernel = {}
    by_range  = {}
    for (v, m) in zip(series.to_numpy(), meta_by_new):
        if m["group"] != "generalizability":
            continue
        if pd.isna(v):
            continue
        tag = "better" if (v < 0) else "worse"
        overall[tag] += 1
        k = m["kernel"]
        r = m["range_label"] or "range=unknown"
        by_kernel.setdefault(k, {"better": 0, "worse": 0})
        by_range.setdefault(r, {"better": 0, "worse": 0})
        by_kernel[k][tag] += 1
        by_range[r][tag]  += 1
    return overall, by_kernel, by_range

def format_title_with_stats(base_title: str, overall, by_kernel, by_range):
    # 总体
    part_overall = f"Genl better={overall['better']} | worse={overall['worse']}"
    # kernel
    ks = [k for k in KERNEL_ORDER if k in by_kernel] + [k for k in by_kernel.keys() if k not in KERNEL_ORDER]
    part_kernel = "; ".join([f"{k}:{by_kernel[k]['better']}/{by_kernel[k]['worse']}" for k in ks]) or "no-genl"
    # range（按名称排序以稳定）
    rs = sorted(by_range.keys())
    part_range  = "; ".join([f"{r}:{by_range[r]['better']}/{by_range[r]['worse']}" for r in rs]) or "no-genl"
    # 拼接（多行标题）
    return f"{base_title}\n{part_overall} | by kernel [{part_kernel}] | by range [{part_range}]"

def plot_one_model(series: pd.Series, meta_by_new, tick_pos, tick_lab, boundaries, title, range_palette):
    n = len(series)
    fig_w = max(MIN_FIG_W, min(MAX_FIG_W, int(n * PX_PER_BAR) + 10))
    fig, ax = plt.subplots(figsize=(fig_w, 6))

    x = np.arange(n)
    y = series.to_numpy(float)

    # 颜色：generalizability 用 range_palette；其他组固定色
    colors = []
    hatches = []
    for v, m in zip(y, meta_by_new):
        if m["group"] == "generalizability":
            label = m["range_label"] or "range=unknown"
            c = range_palette.get(label, get_cmap("Blues")(0.6))
            colors.append(c)
            hatches.append(KERNEL_HATCH.get(m["kernel"], None))
        else:
            colors.append(GROUP_BASE_COLOR.get(m["group"], GROUP_BASE_COLOR["other"]))
            hatches.append(None)

    bars = ax.bar(x, y, color=colors, edgecolor="black", linewidth=0.0)
    for b, h in zip(bars, hatches):
        if h:
            b.set_hatch(h)

    # 0 线、边界、网格
    ax.axhline(0.0, color="black", lw=1.0, alpha=0.8)
    for b in boundaries:
        ax.axvline(b, color="gray", lw=0.6, ls="--", alpha=0.5)

    # x 轴刻度只放四大类
    # 只保留有数据的分组标签
    non_empty_ticks = []
    non_empty_labs  = []
    for pos, lab in zip(tick_pos, tick_lab):
        # 判断该组是否至少有一个样本（通过 meta_by_new）
        if any(m["group"] == lab for m in meta_by_new):
            non_empty_ticks.append(pos)
            non_empty_labs.append(lab)
    ax.set_xticks(non_empty_ticks)
    ax.set_xticklabels(non_empty_labs, fontsize=12)

    # y 轴范围对称
    finite = pd.Series(y).dropna().to_numpy()
    if finite.size:
        yabs = float(np.nanmax(np.abs(finite)))
        ymax = yabs * 1.15 if yabs > 0 else 1.0
    else:
        ymax = 1.0
    ax.set_ylim(-ymax, ymax)
    ax.set_xlim(-0.5, n - 0.5 if n else 0.5)
    ax.set_ylabel("RMSE minus baseline", fontsize=12)

    # 图例：
    # 1) generalizability: range 颜色
    range_labels_in_plot = [m["range_label"] for m in meta_by_new if m["group"] == "generalizability"]
    range_labels_in_plot = [r for r in sorted(set(range_labels_in_plot)) if r]
    legend_elems = []
    for r in range_labels_in_plot:
        legend_elems.append(Patch(facecolor=range_palette.get(r, get_cmap("Blues")(0.6))),
                           )
        legend_elems[-1].set_label(r)

    # 2) kernel 纹理（只列出本图出现的 kernel）
    kernels_in_plot = sorted(set([m["kernel"] for m in meta_by_new if m["group"] == "generalizability"]),
                             key=lambda k: (KERNEL_ORDER.index(k) if k in KERNEL_ORDER else len(KERNEL_ORDER)))
    for k in kernels_in_plot:
        h = KERNEL_HATCH.get(k, None)
        if not h:
            continue
        p = Patch(facecolor="white", edgecolor="black", hatch=h, linewidth=0.0)
        p.set_label(f"kernel={k}")
        legend_elems.append(p)

    # 3) 其他三大类颜色
    for g in ["test", "train", "expanded"]:
        if any(m["group"] == g for m in meta_by_new):
            p = Patch(facecolor=GROUP_BASE_COLOR[g], edgecolor="none", label=g)
            legend_elems.append(p)

    if legend_elems:
        ax.legend(handles=legend_elems, loc="upper left", bbox_to_anchor=(1.01, 1.0), title="Legend")

    # 标题（含统计）
    ax.set_title(title, fontsize=13)

    plt.tight_layout(rect=[0, 0, 0.95, 1])
    return fig, ax

def model_col_iter(pivot_minus: pd.DataFrame):
    """
    迭代所有非 baseline 模型列：
      返回 (col_key, col_title, series)
    """
    if isinstance(pivot_minus.columns, pd.MultiIndex):
        col_mask = np.ones(len(pivot_minus.columns), dtype=bool)
        if "model_group" in pivot_minus.columns.names:
            col_mask &= (pivot_minus.columns.get_level_values("model_group") != "baseline")
        cols = [c for c, m in zip(pivot_minus.columns, col_mask) if m]
        for c in cols:
            s = pivot_minus[c]
            parts = []
            # 用列层级组成友好标题
            for name, val in zip(pivot_minus.columns.names, c if isinstance(c, tuple) else (c,)):
                parts.append(f"{name}={val}")
            title = " | ".join(parts)
            yield c, title, s
    else:
        for c in pivot_minus.columns:
            if str(c).lower() == "baseline":
                continue
            yield c, f"model={c}", pivot_minus[c]

def _format_ratio_str(num, den):
    if den <= 0:
        return "N/A"
    return f"{(num/den)*100:.0f}%"

def plot_all_models_from_pivot(pivot_minus, save_dir=None):
    # Color mapping (blue→red for ranges inside generalizability)
    range_palette = build_generalizability_range_palette(pivot_minus, cmap_name="coolwarm")

    # Row order & layout
    row_order, _, tick_pos, tick_lab, boundaries, meta_by_new, sub_xtick_info = compute_row_order_and_layout(
        pivot_minus, range_palette
    )
    pivot_ord = pivot_minus.iloc[row_order, :]

    # Save dir
    if save_dir is None:
        save_dir = SAVE_DIR
    if save_dir:
        os.makedirs(save_dir, exist_ok=True)

    for col_key, col_title, series in model_col_iter(pivot_ord):
        fig_w = max(20, int(len(series) * 0.06))
        fig, ax = plt.subplots(figsize=(fig_w, 10))

        x = np.arange(len(series))
        y = pd.to_numeric(series, errors="coerce").to_numpy(dtype=float)

        # Colors & hatches (range color + kernel hatch for generalizability; fixed colors for others)
        colors, hatches = [], []
        for m in meta_by_new:
            if m["group"] == "generalizability":
                label = m["range_label"] or "range=unknown"
                colors.append(range_palette.get(label, get_cmap("coolwarm")(0.5)))
                hatches.append(KERNEL_HATCH.get(m["kernel"], None))
            else:
                colors.append(GROUP_BASE_COLOR.get(m["group"], GROUP_BASE_COLOR["other"]))
                hatches.append(None)

        # Bars
        bars = ax.bar(x, y, color=colors, edgecolor="black", linewidth=0.0, align="center")

        # === lock x range to bars and remove margins (critical for alignment) ===
        n = len(series)
        ax.set_xlim(-0.5, n - 0.5)
        ax.margins(x=0)

        # ===== Title with stats (all English) =====
        overall, by_kernel, by_range = summarize_good_bad(series, meta_by_new)
        base_title = f"{col_title} — ΔRMSE (model - baseline)"
        title_en = format_title_with_stats(base_title, overall, by_kernel, by_range)
        ax.set_title(title_en, fontsize=14)

        # Axes & grid
        ax.set_ylabel("ΔRMSE", fontsize=12)
        ax.axhline(0.0, color="black", lw=1.0, alpha=0.8)
        ax.set_axisbelow(True)
        ax.grid(axis='y', linestyle='--', linewidth=1.0, alpha=0.6)

        # Bottom ticks: four major groups (vertical)
        ax.set_xticks(tick_pos)
        ax.set_xticklabels(tick_lab, fontsize=12, rotation=90, ha="center", va="top")
        for b in boundaries:
            ax.axvline(b, color="black", lw=1.0, ls=":", alpha=0.5)

        # Top ticks: (kernel, range) segments WITH Better/Worse counts and percentages (English)
        segs = (sub_xtick_info or {}).get("segments", [])

        # --- count trailing non-generalizability bars (usually test/train/expanded) ---
        trail = 0
        for i in range(len(meta_by_new) - 1, -1, -1):
            if meta_by_new[i]["group"] == "generalizability":
                break
            trail += 1

        if segs:
            top_ticks, top_labels = [], []
            for seg in segs:
                s, e = int(seg["start"]), int(seg["end"])
                block  = y[s:e+1]
                valid  = ~np.isnan(block)
                den    = int(valid.sum())
                better = int((block[valid] < 0).sum())
                worse  = int((block[valid] > 0).sum())
                rb = _format_ratio_str(better, den)
                rw = _format_ratio_str(worse , den)

                rg = seg.get("range_label") or "range=unknown"
                rg_short = re.sub(r"^range=", "", rg)
                lbl = f"{seg['kernel']}\n{rg_short}\nBetter: {better}/{den} ({rb})\nWorse: {worse}/{den} ({rw})"
                top_labels.append(lbl)
                top_ticks.append((s + e) / 2.0)

            # ---- create "ghost" empty ticks over the trailing bars so the top looks blank there ----
            ghost_ticks = []
            if trail > 0:
                start_trail = len(series) - trail
                # 一个 ghost/条（也可以换成只放一个 ghost 在末端的做法）
                ghost_ticks = list(range(start_trail, len(series)))

            ax_top = ax.twiny()

            # ✅ VERY IMPORTANT: keep the SAME xlim as the bottom axis so ticks align with bars
            ax_top.set_xlim(ax.get_xlim())

            # 合并真实顶刻度 + ghost 顶刻度（ghost 标签为空字符串）
            ax_top.set_xticks(top_ticks + ghost_ticks)
            ax_top.set_xticklabels(top_labels + [""] * len(ghost_ticks), rotation=90, fontsize=8)

            ax_top.tick_params(axis='x', pad=2, length=3)
            for sp in ["top", "bottom", "left", "right"]:
                ax_top.spines[sp].set_visible(False)

            # generalizability 内部子块边界（虚线）— 画在主轴上
            inner_boundaries = [segs[i]["start"] - 0.5 for i in range(1, len(segs))]
            for vb in inner_boundaries:
                ax.axvline(vb, color="gray", lw=0.8, ls="--", alpha=0.6)


        # Symmetric y-limits around zero
        finite = pd.Series(y).dropna().to_numpy()
        if finite.size:
            yabs = float(np.nanmax(np.abs(finite)))
            ymax = yabs * 1.15 if yabs > 0 else 1.0
        else:
            ymax = 1.0
        ax.set_ylim(-ymax, ymax)

        # Legend: range colors + kernel hatches + other groups
        legend_elems = []
        range_labels_in_plot = [m["range_label"] for m in meta_by_new if m["group"] == "generalizability"]
        for r in sorted({rl for rl in range_labels_in_plot if rl}):
            p = Patch(facecolor=range_palette.get(r, get_cmap("coolwarm")(0.5)))
            p.set_label(r)
            legend_elems.append(p)
        for k in sorted({m["kernel"] for m in meta_by_new if m["group"] == "generalizability"},
                        key=lambda k: (KERNEL_ORDER.index(k) if k in KERNEL_ORDER else len(KERNEL_ORDER))):
            h = KERNEL_HATCH.get(k)
            if h:
                p = Patch(facecolor="white", edgecolor="black", hatch=h, linewidth=0.0)
                p.set_label(f"kernel={k}")
                legend_elems.append(p)
        for g in ["test", "train", "expanded"]:
            if any(m["group"] == g for m in meta_by_new):
                legend_elems.append(Patch(facecolor=GROUP_BASE_COLOR[g], edgecolor="none", label=g))
        if legend_elems:
            ax.legend(handles=legend_elems, loc="upper left", bbox_to_anchor=(1.01, 1.0), title="Legend")

        plt.tight_layout()
        if save_dir:
            out = os.path.join(save_dir, f"{sanitize_title(col_title)}.png")
            plt.savefig(out, dpi=FIG_DPI, bbox_inches="tight")
        # plt.show()

# ---------- filter: keep family-specific generalizability + ALWAYS keep test/train/expanded ----------
def filter_pivot_by_family_including_others(pivot: pd.DataFrame, family: str) -> pd.DataFrame:
    """
    family: 'GRF', 'LogGRF', 'NegLogGRF'（大小写均可）
    行保留逻辑：
      - generalizability 里：只保留属于该 family 的行
      - test/train/expanded：全部保留
    """
    if not isinstance(pivot.index, pd.MultiIndex) or ("type" not in pivot.index.names):
        raise ValueError("pivot must have a MultiIndex with a 'type' level")

    names = pivot.index.names
    type_lv = pivot.index.get_level_values("type").astype(str).str.lower()

    fam = str(family).lower()
    if fam == "grf":
        mask_family = (
            type_lv.str.startswith("grf_")
            & ~type_lv.str.startswith("loggrf_")
            & ~type_lv.str.startswith("negloggrf_")
        )
    elif fam == "loggrf":
        mask_family = type_lv.str.startswith("loggrf_")
    elif fam == "negloggrf":
        mask_family = type_lv.str.startswith("negloggrf_")
    else:
        raise ValueError("family must be one of: 'GRF', 'LogGRF', 'NegLogGRF'")

    # generalizability 的行（用于和 family 取交集）
    if "dataset_group" in names:
        dg = pivot.index.get_level_values("dataset_group").astype(str).str.lower()
        is_gen = (dg == "generalizability")
        is_others = dg.isin(["test", "train", "expanded"])
    else:
        # 兜底（如果没有 dataset_group），从名字里判断
        if "dataset_name" in names:
            dn = pivot.index.get_level_values("dataset_name").astype(str).str.lower()
        else:
            dn = pd.Index([""] * len(pivot))
        is_gen = dn.str.contains("generalizability")
        is_others = (
            dn.str.contains(r"(?:^|_)test(?:_|$)")
            | dn.str.contains(r"(?:^|_)train(?:_|$)")
            | dn.str.contains("expanded")
        )

    # 选择：该 family 的 generalizability + 所有 test/train/expanded
    mask = (is_gen & mask_family) | is_others
    return pivot.loc[mask]

# ---------- run plots for 3 families; each saved to its own subfolder, ALWAYS with test/train/expanded ----------
def run_family_plots(pivot_minus: pd.DataFrame, save_root: str):
    fam_dirs = {
        "GRF":       os.path.join(save_root, "GRF"),
        "LogGRF":    os.path.join(save_root, "LogGRF"),
        "NegLogGRF": os.path.join(save_root, "NegLogGRF"),
    }
    for fam, out_dir in fam_dirs.items():
        sub = filter_pivot_by_family_including_others(pivot_minus, fam)
        if sub.empty:
            print(f"[skip] {fam}: no rows after filtering.")
            continue
        os.makedirs(out_dir, exist_ok=True)
        print(f"[run] {fam}: {len(sub)} rows → {out_dir}")
        plot_all_models_from_pivot(sub, save_dir=out_dir)

# =================== 用法示例 ===================
# 你的透视表变量名例如 pivot_df（行多级：含 dataset_group/dataset_name/vmin/vmax/type 等；列多级：含 model_group 等）：
# plot_all_models_from_pivot(pivot_minus, save_dir=SAVE_DIR)
run_family_plots(pivot_minus, save_root=SAVE_DIR)
```
