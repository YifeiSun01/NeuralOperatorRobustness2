```python
from pathlib import Path
import csv

def list_pickles(roots):
    """
    roots: list[str|Path] of top-level folders to search
    returns: list of dicts with metadata about each .pkl file
    """
    rows = []
    for root in map(Path, roots):
        for p in root.rglob("*.pkl"):
            rows.append({
                "root": str(root.resolve()),
                "folder_relative_to_root": str(p.parent.relative_to(root)) if p.parent != root else ".",
                "parent_folder_name": p.parent.name,
                "file": p.name,
                "abs_path": str(p.resolve()),
            })
    rows.sort(key=lambda r: (r["root"], r["folder_relative_to_root"], r["file"]))
    return rows

if __name__ == "__main__":
    # <-- replace with your two folders
    roots = ["nu=0.01_PGD_results", "nu=0.0005_PGD_results"]

    results = list_pickles(roots)

    # Pretty print
    for r in results:
        print(f"{r['root']}/{r['folder_relative_to_root']}: {r['file']} -> {r['abs_path']}")
```
