```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the data
file_path = "D:/科研/UIUC research/code/adversarial_robustness_FNO/datasets/1D/Burgers/dim1d_nx1024_N10_solver=exponax_kernel=gaussian_correlation_length0.03_bcperiodic_nu0.005_t1.0_seed45.pt"
data = torch.load(file_path)  # Assuming it's a list/array of tensors

# 2. Extract two samples (e.g., index 0 and 1)
sample1 = data["x"].numpy()  # Convert to NumPy
sample2 = data["y"].numpy()

# 3. Create x-axis (assuming spatial domain [0, 1] with 1024 points)
nx = 1024
x = np.linspace(0, 1, nx)
idx = 3
# 4. Plot both curves
plt.figure(figsize=(10, 5))
plt.plot(x, sample1[idx,:], label="Sample 1", color='blue')
plt.plot(x, sample2[idx,:], label="Sample 2", color='red')

# 5. Customize plot
plt.xlabel("Spatial Domain (x)")
plt.ylabel("Solution Value (u)")
plt.title("1D Burgers Equation Solutions")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
