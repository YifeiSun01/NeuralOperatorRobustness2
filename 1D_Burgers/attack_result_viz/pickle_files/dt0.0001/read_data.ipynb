```python
# %% [markdown]
# Pickle 结构检查器（Notebook 版）
# - 打印各级 key/索引路径、类型，以及 numpy/torch 的 shape/dtype
# - 支持：dict / list / tuple / set / 自定义对象的 __dict__
# - 目标可以是单个 .pkl 文件，或一个目录（默认只遍历该目录下一层 *.pkl）
# - 可调参数：MAX_DEPTH, MAX_ITEMS_PER_LEVEL, SHOW_STATS（是否额外打印 min/max）
# 用法：
#   1) 设置下面的 TARGET（文件或目录）
#   2) 运行本 cell


# %%
import os
import pickle
import numbers
from pathlib import Path
from typing import Any

# 可选依赖
try:
    import numpy as np
except Exception:
    np = None

try:
    import torch
except Exception:
    torch = None


# ---------- 参数区（按需改动） ----------
TARGET = "gradient_test_exponax_nu0.0005_nsamples1_2.pkl"   # ←←← 改成你的 .pkl 文件或目录
MAX_DEPTH = 12                             # 最大递归深度
MAX_ITEMS_PER_LEVEL = 50                   # 每一层最多展示的元素数
SHOW_STATS = False                         # 对数组/张量是否额外打印 min/max（可能稍慢）
RECURSE_DIR = False                        # 如果 TARGET 是目录，是否递归子目录
# --------------------------------------


def is_numpy_array(x: Any) -> bool:
    return (np is not None) and isinstance(x, np.ndarray)


def is_torch_tensor(x: Any) -> bool:
    return (torch is not None) and isinstance(x, torch.Tensor)


def short_repr(x: Any, limit: int = 120) -> str:
    try:
        r = repr(x)
    except Exception:
        r = f"<unrepr {type(x).__name__}>"
    return (r[:limit - 3] + "...") if len(r) > limit else r


def describe_leaf(x: Any, do_stats: bool = False) -> str:
    """返回叶子节点的简要描述（含数组/张量的 shape/dtype）。"""
    try:
        if is_numpy_array(x):
            base = f"numpy.ndarray shape={tuple(x.shape)} dtype={x.dtype}"
            if do_stats:
                # 使用 nanmin / nanmax 避免 NaN 干扰
                try:
                    vmin = float(np.nanmin(x)) if x.size else None
                    vmax = float(np.nanmax(x)) if x.size else None
                    base += f" min={vmin} max={vmax}"
                except Exception:
                    pass
            return base
        if is_torch_tensor(x):
            base = f"torch.Tensor shape={tuple(x.shape)} dtype={x.dtype} device={x.device}"
            if do_stats and x.numel() > 0:
                try:
                    vmin = float(torch.nanmin(x)) if hasattr(torch, "nanmin") else float(x.min())
                    vmax = float(torch.nanmax(x)) if hasattr(torch, "nanmax") else float(x.max())
                    base += f" min={vmin} max={vmax}"
                except Exception:
                    pass
            return base
        if isinstance(x, (bytes, bytearray)):
            return f"{type(x).__name__}(len={len(x)})"
        if isinstance(x, str):
            return f"str(len={len(x)})"
        if isinstance(x, numbers.Number):
            return f"{type(x).__name__}({x})"
        return type(x).__name__
    except Exception as e:
        return f"{type(x).__name__}  <describe_error: {e}>"


def iter_mapping(m):
    """安全遍历 mapping 类型（dict 等），限制个数。"""
    try:
        items = list(m.items())
    except Exception:
        return []
    return items[:MAX_ITEMS_PER_LEVEL]


def iter_sequence(seq):
    """安全遍历序列类型（list/tuple/set），限制个数。"""
    try:
        if isinstance(seq, set):
            seq = list(seq)
        return list(enumerate(seq))[:MAX_ITEMS_PER_LEVEL]
    except Exception:
        return []


def walk(obj: Any, path: str = "root", depth: int = 0):
    """递归打印结构。"""
    indent = "  " * depth
    if depth > MAX_DEPTH:
        print(f"{indent}{path}: <max depth reached>")
        return

    # dict / mapping
    if hasattr(obj, "items") and callable(getattr(obj, "items", None)):
        print(f"{indent}{path}: dict(len={len(obj)})")
        for k, v in iter_mapping(obj):
            key_str = short_repr(k, limit=80)
            walk(v, f"{path}[{key_str}]", depth + 1)
        if len(obj) > MAX_ITEMS_PER_LEVEL:
            print(f"{indent}  ... ({len(obj) - MAX_ITEMS_PER_LEVEL} more keys not shown)")
        return

    # list / tuple / set
    if isinstance(obj, (list, tuple, set)):
        tname = type(obj).__name__
        try:
            ln = len(obj)
        except Exception:
            ln = "?"
        print(f"{indent}{path}: {tname}(len={ln})")
        for idx, v in iter_sequence(obj):
            walk(v, f"{path}[{idx}]", depth + 1)
        if isinstance(obj, (list, tuple, set)) and isinstance(ln, int) and ln > MAX_ITEMS_PER_LEVEL:
            print(f"{indent}  ... ({ln - MAX_ITEMS_PER_LEVEL} more items not shown)")
        return

    # numpy / torch / 叶子
    desc = describe_leaf(obj, do_stats=SHOW_STATS)
    print(f"{indent}{path}: {desc}")

    # 对于自定义对象，补充打印 __dict__（若有）
    try:
        if hasattr(obj, "__dict__") and isinstance(obj.__dict__, dict) and len(obj.__dict__) > 0:
            print(f"{indent}  {path}.__dict__: dict(len={len(obj.__dict__)})")
            for k, v in iter_mapping(obj.__dict__):
                walk(v, f"{path}.__dict__[{short_repr(k)}]", depth + 1)
    except Exception:
        pass


def load_pickle_safely(pkl_path: Path):
    """尽可能兼容地加载 pickle。"""
    # 常见的跨环境问题：Py2/latin1；或使用不同协议
    tried = []
    try:
        with open(pkl_path, "rb") as f:
            return pickle.load(f)
    except Exception as e:
        tried.append(str(e))
    try:
        with open(pkl_path, "rb") as f:
            return pickle.load(f, encoding="latin1")
    except Exception as e:
        tried.append(str(e))
    raise RuntimeError(f"Failed to load {pkl_path}.\nTried errors:\n- " + "\n- ".join(tried))


def inspect_one_pickle(pkl_path: Path):
    print(f"\n[LOAD] {pkl_path}")
    obj = load_pickle_safely(pkl_path)
    walk(obj)


def main(target: str):
    p = Path(target)
    if p.is_file() and p.suffix.lower() == ".pkl":
        inspect_one_pickle(p)
        return

    if p.is_dir():
        pattern = "**/*.pkl" if RECURSE_DIR else "*.pkl"
        files = sorted(p.glob(pattern))
        print(f"Found {len(files)} pickle files in {p}:")
        for f in files:
            print("  -", f.name)
        for f in files:
            inspect_one_pickle(f)
        return

    raise FileNotFoundError(f"TARGET not found or not a .pkl/dir: {p}")


# 运行
try:
    main(TARGET)
except Exception as e:
    print("[ERROR]", e)
```

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from typing import Optional, Mapping, Any

def _to_1d(x: Any) -> np.ndarray:
    """Convert tensor/array/list to 1D numpy array (auto-squeeze)."""
    if isinstance(x, torch.Tensor):
        x = x.detach().cpu().numpy()
    x = np.asarray(x)
    x = np.squeeze(x)         # e.g. (1, N, 1) -> (N,)
    if x.ndim != 1:
        # fallback：展平成一维（通常 squeeze 后已是一维）
        x = x.reshape(-1)
    return x.astype(float)

def plot_T_and_G(
    sample: Mapping[str, Any],
    title: Optional[str] = "T_out vs G_out",
    save_path: Optional[str] = None,
    show: bool = True,
):
    """
    画两条折线：T_out（真值/目标）与 G_out（模型输出）。

    Parameters
    ----------
    sample : dict-like
        至少包含键 'T_out' 与 'G_out'，它们可以是 numpy 数组、torch 张量或可转为数组的对象。
        典型形状：G_out ~ (1, N, 1) 或 (N,)，T_out ~ (N,)
    title : str, optional
        图标题。
    save_path : str, optional
        若提供，则把图保存到该路径（例如 "tg_plot.png"）。
    show : bool, default True
        是否调用 plt.show() 显示图像（在脚本/无界面环境可设为 False）。
    """
    if "G_out" not in sample or "T_out" not in sample:
        raise KeyError("sample 必须包含 'G_out' 和 'T_out' 两个键。")

    G = _to_1d(sample["G_out"])
    T = _to_1d(sample["T_out"])

    # 对齐长度（若两者长度不一致，取较短部分）
    n = min(len(G), len(T))
    G, T = G[:n], T[:n]
    x = np.arange(n)

    plt.figure(figsize=(8, 4.5))
    plt.plot(x, T, linewidth=2, label="T_out")
    plt.plot(x, G, linewidth=2, linestyle="--", label="G_out")
    plt.xlabel("Time Index")
    plt.ylabel("Value")
    plt.title(title)
    plt.legend()
    plt.grid(True, linestyle=":")
    plt.tight_layout()

    if save_path is not None:
        plt.savefig(save_path, dpi=150)

    if show:
        plt.show()
    else:
        plt.close()
```
