```python
# Training data (first 1400 samples)
x_train = data["x"][:1400, :]  # Shape: [1400, 1024]
y_train = data["y"][:1400, :]  # Shape: [1400, 1024]

# Testing data (last 100 samples)
x_test = data["x"][1400:, :]   # Shape: [100, 1024]
y_test = data["y"][1400:, :]    # Shape: [100, 1024]
```

```python
# Save training data (with "_train" suffix)
torch.save({
    "x": x_train,
    "y": y_train,
}, "dim1d_nx1024_N1400_train.pt")  # You can adjust the filename

# Save testing data (with "_test" suffix)
torch.save({
    "x": x_test,
    "y": y_test,
}, "dim1d_nx1024_N100_test.pt")  # You can adjust the filename
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
idx = 5
x = np.linspace(0, 1, 100)  # 100 points from 0 to 10
y = d["X"][0][idx]                # Sine wave
z = d["X"][1]
w = d["y"][idx]

# Create the plot
plt.figure(figsize=(8, 4))     # Set figure size
plt.plot(x, y, label='', color='blue', linewidth=2)  # Line plot
plt.plot(x, z, label='', color='red', linewidth=2)  # Line plot
plt.plot(x, w, label='', color='brown', linewidth=2)  # Line plot

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Line Plot')
plt.legend()  # Show legend
plt.grid(True)  # Add grid

# Display the plot
plt.show()
```
