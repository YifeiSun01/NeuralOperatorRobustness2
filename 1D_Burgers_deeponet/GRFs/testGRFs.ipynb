```python
from generateGRFs import GRFGenerator, visualize_GRF
import numpy as np

grf_params = {
    'dim': 1,
    'nx': 2**11,
    'kernel': 'matern',
    'kernel_params': {'correlation_length': 0.03, "nu":2.5},
    'bc': 'periodic',
    'seed': 45,
    'zero_mean': False
}

grf = GRFGenerator.generate_grf(
    shape=(int(grf_params["nx"]),),  # Ensure tuple of integers
    kernel=grf_params['kernel'],
    kernel_params=grf_params['kernel_params'],
    bc=grf_params['bc'],
    seed=int(grf_params['seed']),
    zero_mean=grf_params['zero_mean']
) # + np.random.normal(0, 0.01, grf_params["nx"])

visualize_GRF(grf)
```
