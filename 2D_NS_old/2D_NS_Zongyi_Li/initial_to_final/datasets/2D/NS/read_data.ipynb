```python
import numpy as np
import matplotlib.pyplot as plt

def plot_dual_heatmaps(array1, array2, 
                      title1='Heatmap 1', title2='Heatmap 2',
                      cmap='viridis', figsize=(12, 6),
                      vmin=None, vmax=None,
                      colorbar=True,idx=0):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=figsize)
    if vmin is None:
        vmin = min(np.nanmin(array1), np.nanmin(array2))
    if vmax is None:
        vmax = max(np.nanmax(array1), np.nanmax(array2))
    im1 = ax1.imshow(array1, cmap=cmap, vmin=vmin, vmax=vmax, origin='lower')
    ax1.set_title(title1)
    ax1.set_xlabel('X')
    ax1.set_ylabel('Y')
    if colorbar:
        fig.colorbar(im1, ax=ax1, label='Value')
    im2 = ax2.imshow(array2, cmap=cmap, vmin=vmin, vmax=vmax, origin='lower')
    ax2.set_title(title2)
    ax2.set_xlabel('X')
    if colorbar:
        fig.colorbar(im2, ax=ax2, label='Value')
    ax2.set_ylabel('')
    plt.suptitle(f"Navier-Stokes Vorticity Transport idx={idx}")
    plt.tight_layout()
    plt.show()
```

```python
import torch

data = torch.load('NS_data_zongyi_train.pt', weights_only=False)
```

```python
import numpy as np

for idx in range(5):
    plot_dual_heatmaps(data["x"][idx], data["y"][idx],title1='input', title2='output', idx=idx+1)
```
